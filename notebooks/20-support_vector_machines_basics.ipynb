Verify critical information before relying on it.# Support Vector Machines (Basics)A support vector machine finds a decision boundary with the largest margin between classes.

## Core IdeaSoft-margin SVMs balance margin width and violations through $C$. Kernels replace inner products to create nonlinear boundaries without explicitly constructing all transformed features.

```python
import numpy as np
from sklearn.datasets import make_moons
from sklearn.svm import SVC

X, y = make_moons(n_samples=200, noise=0.15, random_state=42)
linear = SVC(kernel='linear').fit(X, y)
rbf = SVC(kernel='rbf', gamma='scale').fit(X, y)

print('linear accuracy:', linear.score(X, y))
print('RBF accuracy:', rbf.score(X, y))
print('RBF support vectors:', len(rbf.support_))
```

## Key TakeawaysOnly support vectors determine the fitted boundary. Kernel and regularization choices control flexibility.
