## Core IdeaThe E-step computes responsibilities using current parameters. The M-step updates parameters using those soft assignments. Each iteration does not decrease the data likelihood.

```python
import numpy as nprng = np.random.default_rng(7)x = np.r_[rng.normal(-2, 0.6, 150), rng.normal(2.5, 0.8, 150)]means = np.array([-1.0, 1.0]); stds = np.ones(2); weights = np.ones(2) / 2for _ in range(30):    density = np.exp(-0.5 * ((x[:, None] - means) / stds) ** 2) / stds    responsibility = density * weights    responsibility /= responsibility.sum(axis=1, keepdims=True)    mass = responsibility.sum(axis=0)    weights = mass / len(x)    means = (responsibility * x[:, None]).sum(axis=0) / mass    stds = np.sqrt((responsibility * (x[:, None] - means) ** 2).sum(axis=0) / mass)print('means:', means.round(2), 'stds:', stds.round(2))
```
